Machine Learning pipeline


```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Import machine learning modules from scikit-learn
from sklearn import metrics
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import f_classif
```

## **2. Load and Understand the dataset**

```python
# Convert to DataFrame
df = pd.read_csv('UCI_Wine_Dataset_178.csv')
```

## **3. Visualize the dataset**


```python
plt.hist(df['Alcohol'])
plt.xlabel("Alcohol")
plt.ylabel("Number of Rows")
plt.show()
```

```python
plt.hist(df['Malicacid'])
plt.xlabel("Malicacid")
plt.ylabel("Number of Rows")
plt.show()
```

```python
plt.scatter(df['Alcohol'], df['Malicacid'], c=df['class'].map(colors))
plt.xlabel("Alcohol")
plt.ylabel("Malicacid")
plt.show()
```

```python
#features
x = df.drop(columns=["class", "Id"]).values

#labels
y = df["class"].values
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
print(x_train)
```

Create a K-Nearest Neighbors classifier

```python
knn = KNeighborsClassifier(n_neighbors=13)
```

Train the model using the training data

```python
knn.fit(x_train, y_train)
```

Compute and display model accuracy

```python
y_pred = knn.predict(x_test)
metrics.accuracy_score(y_test, y_pred)
```

```python
knn.predict([[1.24,3.22,3.52,7.22,5.25,3.62,7.25,9.22,9.61,11.23,11.93,15.22,13.62]])
```

Display the confusion matrix

```python
metrics.confusion_matrix(y_test, y_pred)
```

4. Feature selection

```python
X = df.drop(["class", "Id"], axis=1)
y = df["class"]
```

```python
F, p = f_classif(X, y)

feature_scores = pd.DataFrame({
    "Feature": X.columns,
    "F_Score": F,
    "p_value": p
})

feature_scores = feature_scores.sort_values(
    "F_Score",
    ascending=False
)

feature_scores
```

```python
top_features = feature_scores.head(5)
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    top_features["Feature"],
    top_features["F_Score"]
)

plt.xticks(rotation=45)
plt.xlabel("Feature")
plt.ylabel("F-Score")
plt.title("Top 5 Features")
plt.show()
```

Split the dataset: 80% training and 20% testing

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=5
)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

Predict the classes of the test data

```python
y_pred = knn.predict(X_test)
```

Accuracy

```python
accuracy = metrics.accuracy_score(y_test, y_pred)
print('Accuracy:', accuracy)
```

Precision

```python
precision = metrics.precision_score(y_test, y_pred, average='weighted')
print('\nPrecision:', precision)
```

Recall

```python
recall = metrics.recall_score(y_test, y_pred, average='weighted')
print('\nRecall:', recall)
```

F1-Score

```python
f1 = metrics.f1_score(y_test, y_pred, average='weighted')
print('\nF1-Score:', f1)
```
